```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
doc1 = Document(
    page_content="Total Gaming is a Free Fire gaming team known for aggressive gameplay, coordinated pushes, and close-range fights.",
    metadata={"team": "Total Gaming", "style": "Aggressive"}
)

doc2 = Document(
    page_content="GodLike Esports is an esports organization associated with competitive gaming. Team coordination, strategy, and tournament performance are important aspects of its gameplay.",
    metadata={"team": "GodLike Esports", "style": "Competitive"}
)

doc3 = Document(
    page_content="Team Elite is associated with the Indian Free Fire competitive gaming community. Its team strategy includes coordinated fights and tactical positioning.",
    metadata={"team": "Team Elite", "style": "Tactical"}
)

doc4 = Document(
    page_content="4 Unknown is a Free Fire esports team known for its competitive presence. Teamwork, rotations, and survival are important in battle royale matches.",
    metadata={"team": "4 Unknown", "style": "Strategic"}
)

doc5 = Document(
    page_content="LOUD is an esports organization that has competed in Free Fire tournaments. Its competitive gameplay involves team coordination, combat strategy, and positioning.",
    metadata={"team": "LOUD", "style": "Competitive"}
)
```

```python
docs=[doc1,doc2,doc3,doc4,doc5]
print("Total documents:",len(docs))
```

```python
embedings=HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
vector_store=Chroma(
    embedding_function=embedings,
    persist_directory="./freefire_chroma_db",
    collection_name="freefire_teams"
)
```

```python
# Add Documents
vector_store.add_documents(
    documents=docs,
    ids=[
        "total-gaming",
        "godlike",
        "team-elite",
        "4-unknown",
        "loud"
    ]
)
```

```python
# View Stored Documents
vector_store.get(
    include=["documents","metadatas","embeddings"]
)
```

```python
# Similarity Search
results=vector_store.similarity_search(
    query="which team uses aggressive gamepay?",
    k=2
)
for doc in results:
  print(doc.page_content)
  print(doc.metadata)
  print("---")
```

```python
# Similarity Search With Score
results=vector_store.similarity_search_with_score(
    query="which team gameplay is competative?",
    k=2
)
for doc,score in results:
  print("team:",doc.metadata["team"])
  print("Content",doc.page_content)
  print("DISTANCE:",score)
  print("---")
```

```python
# Metadata Filtering
results=vector_store.similarity_search(
    query="Competitive gameplay",
    k=2,
    filter={"style":"Competitive"}
)
for doc in results:
  print(doc.page_content)
  print(doc.metadata)
  print("---")
```

```python
from chromadb import Metadata
# Update a Document
updated_doc = Document(
    page_content="Total Gaming is a Free Fire team focused on aggressive pushes, coordinated squad attacks, fast rotations, and strategic positioning during competitive matches.",
    metadata={"team": "Total Gaming", "style": "Aggressive"}
)

vector_store.update_document(
    document_id="total-gaming",
    document=updated_doc
)
```

```python
# Verify Update
vector_store.get(
    ids=["total-gaming"],
    include=["documents","metadatas"]
)
```

```python
# Delete a Document
vector_store.delete(
    ids=["4-unknown"]
)
```

```python
# Verify Deletion
vector_store.get(
    include=["documents","metadatas"]
)
```

```python
# Test Search Again
results=vector_store.similarity_search(
    query="Tell me about Free Fire teams",
    k=4
)
for doc in results:
  print(doc.page_content)
  print(doc.metadata["team"])
  print("---")
```
